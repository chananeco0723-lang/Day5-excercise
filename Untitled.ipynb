```python
import pandas as pd

# ===== 1. 读入四个 CSV =====
# 注意：列名是 observation_date，不是 DATE
gs10 = pd.read_csv("data/GS10.csv", parse_dates=["observation_date"], index_col="observation_date")
mortgage = pd.read_csv("data/MORTGAGE30US.csv", parse_dates=["observation_date"], index_col="observation_date")
hsales = pd.read_csv("data/HSN1F.csv", parse_dates=["observation_date"], index_col="observation_date")
cpi = pd.read_csv("data/CPIAUCSL.csv", parse_dates=["observation_date"], index_col="observation_date")

# ===== 2. 把列名改成简单好懂的 =====
gs10.columns = ["US_10Y_Treasury"]
mortgage.columns = ["US_30Y_Mortgage"]
hsales.columns = ["US_New_Home_Sales"]
cpi.columns = ["US_CPI"]

# ===== 3. 合并成一张大表 =====
macro = pd.concat([gs10, mortgage, hsales, cpi], axis=1).dropna()

# ===== 4. 看看结果 =====
print("数据合并成功！前 3 行：")
display(macro.head(3))
```

```python
# ===== 5. 保存数据供 Streamlit 使用 =====
macro.to_csv("data/macro_monthly.csv")
print("✅ 已保存到 data/macro_monthly.csv")
```
